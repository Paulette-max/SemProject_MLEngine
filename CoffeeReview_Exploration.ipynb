```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('coffee_clean.csv')

print("Dataset loaded successfully!")
```

```python
# How many rows and columns are contained in the dataset?
num_rows, num_cols = df.shape

print(f"The dataset has {num_rows} rows and {num_cols} columns.")
```

```python
# What datatypes are contained in the dataset?
print(df.dtypes)
```

```python
# c. Is the dataset complete?
missing_values = df.isnull().sum()
total_missing = missing_values.sum()

print(f"\nTotal missing values: {total_missing}")
if total_missing == 0:
    print("The dataset is complete with no missing values.")
else:
    print("Missing values per column:")
    print(missing_values[missing_values > 0])
```

```python
# d. Slice out the first 15 rows and last 20 rows and merge them into a new dataframe titled “df_sample”.
first_15 = df.head(15)
last_20 = df.tail(20)

df_sample = pd.concat([first_15, last_20], ignore_index=True)

print(f"Sample DataFrame Shape: {df_sample.shape}")
display(df_sample)
```
